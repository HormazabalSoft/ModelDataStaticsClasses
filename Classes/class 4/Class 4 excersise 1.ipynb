```python
import os
import pandas as pd
import math
import matplotlib.pyplot as plt
```

```python


Hours = [2,4,6,8,10]
Marks = [45,55,65,70,80]

df = pd.DataFrame({
        "Hours": Hours,
        "Marks": Marks
    })

```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
#use Linear Regression
modelo = LinearRegression()
```

```python
X = df["Hours"].values.reshape(-1,1)
Y = df["Marks"].values
```

```python
#transform
modelo.fit(X, Y)
```

### Answer 1:

```python
#print the values B1, B0 and R2
print("Pendiente (β₁):", modelo.coef_[0])
print("Intercepto (β₀):", modelo.intercept_)
```

### Answer 2:

```python
x2 = 7
y2 = modelo.coef_[0]*x2 + modelo.intercept_
print("Y:", y2)
```

When x is 7, Y is 67.25

### Answer 3:

```python
#predict values
Y_pred = modelo.predict(X)
df["y_pred"] = Y_pred
df["y_pred"]
```

```python
residuos = Y - Y_pred
df["residuos"] = residuos
df["residuos"]
```

Resudals values are very low, so values are near zero.

### Answer 4:

```python
print("R²:", modelo.score(X, Y))
```

The R2 es aproximately 98,9%, so predictions are highly reliable, becase R2 are too high and residuals are very small

### Answer 5:

The relationship is highly reliable because the model has a very high R
2 value and the residuals are small, indicating accurate and consistent predictions.
