```python
import os
import pandas as pd
import math
import matplotlib.pyplot as plt
```

```python

Hours = [4,5,6,7,8]
productivity = [48,52,60,68,75]

df = pd.DataFrame({
        "Hours": Hours,
        "Productivity": productivity
    })
```

```python
X = df["Hours"].values.reshape(-1,1)
Y = df["Productivity"].values
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
#use Linear Regression
modelo = LinearRegression()

#transform
modelo.fit(X, Y)
```

### Answer 1:

```python
#print the values B1, B0 and R2
print("Pendiente (β₁):", modelo.coef_[0])
print("Intercepto (β₀):", modelo.intercept_)
```

The relationship between sleep and productivity can be modeled using a linear regression equation. The slope 7 indicates that for each aditional hour of sleep, productivity increases by 7 units on average. The intercept 18.6 represents the predicted productivity when sleep is zero.

### Answer 2:

```python
# Filtramos primero los datos de entrada
x2 = 6.5
y2 = modelo.coef_[0]*x2 + modelo.intercept_
print("Y:", y2)
```

When hours is 6.5, the productivity will be is 64.1
